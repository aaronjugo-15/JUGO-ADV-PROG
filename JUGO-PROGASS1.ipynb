## A. Word Rotation Problem

```python
print("Word Rotation Problem \n ")

def rotate_word(text):
    x = text.replace(" ","")
    t = x[1:] + x[0]
    return t

text = input("\nInput your own word: ")

rotate_word(text)
```

```python
rotate_word("python") 
```

```python
rotate_word("logic")
```

```python
rotate_word("Code")
```

```python
rotate_word("A")
```

## B.  USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):
    x = first_name.lower()
    y = last_name.lower()
    return (x.replace(" ", "") + "." + y.replace(" ", ""))
    
print("--- Examples---")
exp_f_name = "Aaron Siegfreid"
exp_l_name = "Jugo"
s = exp_f_name.lower()
d = exp_l_name.lower()
print(exp_f_name +" "+ exp_l_name + " --> " + s.replace(" ", "") + "." + d.replace(" ", ""))


first_name = input("\nInput your first name: ")
last_name = input("Input your last name: ")

make_username(first_name, last_name)
```

```python
make_username("Ada", "Lovelace")
```

```python
make_username("Alan", "Turing")
```

```python
make_username("Ana Maria", "De Leon")
```

## C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    first = items[-1]
    middle = items[1:-1:1]
    last = items[-0]
    items = first, *middle, last 
    return items 

text = input("Input your list:")
items = text.split(",") 

swap_bookends(items)
```

```python
swap_bookends([1, 2, 3, 4, 5, 6]) 
```

```python
swap_bookends(["red", "green", "blue"])
```

```python
swap_bookends([8, 3]) 
```
